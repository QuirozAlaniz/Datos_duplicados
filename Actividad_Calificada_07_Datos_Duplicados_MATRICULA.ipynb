### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
# Escribe tu código aquí
hoteles2 = hoteles.copy()                              #Hacemos una copia del dataset original y lo guardamos en "hoteles2"
```

```python
hoteles2["es_duplicado"] = hoteles.duplicated(subset = None, keep = "first")                 #Agregamos la columna booleana a los duplicados sin contar primeras ocurrencias
agrupacion = hoteles2[["hotel", "es_duplicado"]].groupby("hotel", as_index = False).agg({"es_duplicado":"sum"})     #El "as_index = False" evita que la columna hotel
                                                                                                                    #se vuelva indice
duplicados_city = agrupacion.iloc[0, 1]                                                                             #Accedemos al numero de duplicados del City Hotel
duplicados_resort = agrupacion.iloc[1, 1]                                                                           #Accedemos al número de duplicados del Resort Hotel
print(f'Los duplicados del City Hotel son: {duplicados_city}')
print()
print(f'Los duplicados del Resort Hotel son: {duplicados_resort}')

reservas_city = hoteles2[hoteles2["hotel"] == "City Hotel"]                                                  #Obtenemos el total de reservas de City Hotel

reservas_resort = hoteles2[hoteles2["hotel"] == "Resort Hotel"]                                                #Obtenemos el total de reservas de Resort Hotel

print()
print(f'El porcentaje de registros duplicados en City Hotel es de {duplicados_city/len(reservas_city):.2%}')
print()
print(f'El porcentaje de registros duplicados en Resort Hotel es de {duplicados_resort/len(reservas_resort):.2%}')
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
# Escribe tu código aquí
#Hacemos los cálculos para el dataframe original.

hoteles = hoteles_original.drop("duplicados_first", axis = 1)

agrupacion = hoteles.groupby(by = "hotel").size()

cancelacion = hoteles.groupby(by = "hotel")["is_canceled"].mean()*100

tarifa_promedio = hoteles.groupby("hotel")["adr"].mean()

#Hacemos los cálculos para el dataframe limpio.
agrupacion_limpio = hoteles_limpio.groupby("hotel").size()

cancelacion_limpio = hoteles_limpio.groupby("hotel")["is_canceled"].mean()*100

tarifa_promedio_limpio = hoteles_limpio.groupby("hotel")["adr"].mean()

#Creamos las tablas como dataframes para mostrar los resultados
resumen_original = pd.DataFrame({"hotel":agrupacion.index, "tasa_cancelacion":cancelacion.values, "tarifa_diaria_promedio": tarifa_promedio.values})
print(f'El resumen del dataframe original es: \n {resumen_original}')
print()
resumen_limpio = pd.DataFrame({"hotel": agrupacion_limpio.index, "tasa_cancelacion":cancelacion_limpio.values , "tarifa_diaria_promedio":tarifa_promedio_limpio.values })
print(f'El resumen del dataframe limpio es: \n {resumen_limpio}')
```

```python
comparacion = pd.concat([resumen_original, resumen_limpio], ignore_index = True)       #Ponemos un dataframe encima del otro e ignoramos el índice.
comparacion = comparacion.sort_values(by = "hotel").reset_index(drop = True)           #Reordenamos por hotel y reseteamos el índice.
print(comparacion)
```
